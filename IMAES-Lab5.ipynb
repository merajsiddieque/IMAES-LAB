```python
# Experiment 4 : Backward Chaining Engine

# Rules (IF conditions THEN fault)

rules = {
    "dead_battery": {
        "conditions": ["lights_dim", "engine_not_start"]
    },

    "starter_motor_fault": {
        "conditions": ["clicking_sound", "engine_not_cranking"]
    },

    "fuel_problem": {
        "conditions": ["engine_cranks", "fuel_empty"]
    }
}

# Questions for askable facts
questions = {
    "lights_dim": "Are the headlights dim? (yes/no/why): ",
    "engine_not_start": "Does the engine fail to start? (yes/no/why): ",
    "clicking_sound": "Do you hear a clicking sound? (yes/no/why): ",
    "engine_not_cranking": "Is the engine not cranking? (yes/no/why): ",
    "engine_cranks": "Does the engine crank? (yes/no/why): ",
    "fuel_empty": "Is the fuel tank empty? (yes/no/why): "
}
```

```python
facts = {}

def ask(fact, current_rule):
    """Ask the user for a fact"""

    if fact in facts:
        return facts[fact]

    while True:
        ans = input(questions[fact]).strip().lower()

        if ans == "why":
            print(f"\nWHY? Because I am trying to verify the rule for '{current_rule}'.")
            print(f"Rule: IF {', '.join(rules[current_rule]['conditions'])} THEN {current_rule}\n")

        elif ans == "yes":
            facts[fact] = True
            return True

        elif ans == "no":
            facts[fact] = False
            return False

        else:
            print("Enter only yes, no, or why.")


def prove(goal):
    """Try to prove one hypothesis"""

    conditions = rules[goal]["conditions"]

    for fact in conditions:
        if not ask(fact, goal):
            return False

    return True


def diagnose(hypothesis_order):
    """Check hypotheses one by one"""

    global facts
    facts = {}

    print("\n===== Car Fault Diagnosis =====\n")

    for goal in hypothesis_order:
        if prove(goal):
            print("\nDiagnosis:", goal.replace("_", " ").title())
            return

    print("\nDiagnosis: Unknown fault")
```

```python
order1 = [
    "dead_battery",
    "starter_motor_fault",
    "fuel_problem"
]

diagnose(order1)
```

```python
order2 = [
    "starter_motor_fault",
    "dead_battery",
    "fuel_problem"
]

diagnose(order2)
```

```python
diagnose(order2)
```
