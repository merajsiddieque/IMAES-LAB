```python
# ==========================================
# Experiment 1
# Semantic Network with Inheritance
# ==========================================

# Knowledge Base (Subject, Relation, Object)
knowledge = [
    ("penguin", "is-a", "bird"),
    ("bird", "is-a", "animal"),
    ("fish", "is-a", "animal"),
    ("shark", "is-a", "fish"),

    ("animal", "has", "skin"),
    ("bird", "can", "fly"),

    ("penguin", "cannot", "fly")
]


def parent(node):
    for s, r, o in knowledge:
        if s == node and r == "is-a":
            return o
    return None


def inheritance_chain(node):
    chain = []
    while node:
        chain.append(node)
        node = parent(node)
    return chain


def query(node, relation):
    chain = inheritance_chain(node)

    for current in chain:

        # Check exception first
        if relation == "can":
            for s, r, o in knowledge:
                if s == current and r == "cannot":
                    return f"cannot {o}"

        # Check property
        for s, r, o in knowledge:
            if s == current and r == relation:
                return o

    return "unknown"


print("=" * 50)
print("SEMANTIC NETWORK")
print("=" * 50)

print("\nKnowledge Base\n")
for triple in knowledge:
    print(triple)

print("\nQueries\n")

tests = [
    ("penguin", "can"),
    ("bird", "can"),
    ("shark", "has"),
    ("penguin", "has"),
]

for node, rel in tests:
    print(f"Query({node}, {rel}) -> {query(node, rel)}")

print("\nInheritance Chain")
print(" -> ".join(inheritance_chain("penguin")))
```

```python
# ==========================================
# Exercise 2
# Frame Representation
# ==========================================

frames = {

    "animal": {
        "ako": None,
        "has": "skin"
    },

    "bird": {
        "ako": "animal",
        "can": "fly"
    },

    "penguin": {
        "ako": "bird",
        "cannot": "fly"
    },

    "fish": {
        "ako": "animal"
    },

    "shark": {
        "ako": "fish"
    }
}


def frame_query(node, slot):

    while node:

        frame = frames[node]

        if slot == "can" and "cannot" in frame:
            return "cannot " + frame["cannot"]

        if slot in frame:
            return frame[slot]

        node = frame["ako"]

    return "unknown"


print("=" * 50)
print("FRAME REPRESENTATION")
print("=" * 50)

print("Penguin can :", frame_query("penguin", "can"))
print("Bird can    :", frame_query("bird", "can"))
print("Shark has   :", frame_query("shark", "has"))
```

```python
# ==========================================
# Exercise 3
# Multiple Inheritance
# ==========================================

knowledge2 = [

    ("bat", "is-a", "mammal"),
    ("bat", "is-a", "flying_animal"),

    ("mammal", "has", "hair"),
    ("flying_animal", "can", "fly")
]

parents = {}

for s, r, o in knowledge2:
    if r == "is-a":
        parents.setdefault(s, []).append(o)


def search(node, relation, visited=None):

    if visited is None:
        visited = set()

    if node in visited:
        return "unknown"

    visited.add(node)

    # Check local property
    for s, r, o in knowledge2:
        if s == node and r == relation:
            return o

    # Search all parents
    for p in parents.get(node, []):
        ans = search(p, relation, visited)
        if ans != "unknown":
            return ans

    return "unknown"


print("=" * 50)
print("MULTIPLE INHERITANCE")
print("=" * 50)

print("Bat can :", search("bat", "can"))
print("Bat has :", search("bat", "has"))
```
