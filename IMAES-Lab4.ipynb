```python
# Experiment 3: Forward Chaining Inference Engine

# IF-THEN Production Rules

rules = [
    {"id": "R1", "conditions": ["has hair"], "conclusion": "is mammal"},
    {"id": "R2", "conditions": ["gives milk"], "conclusion": "is mammal"},
    {"id": "R3", "conditions": ["eats meat"], "conclusion": "is carnivore"},
    {"id": "R4", "conditions": ["is mammal", "is carnivore"], "conclusion": "is lion"},
    {"id": "R5", "conditions": ["is mammal", "black stripes"], "conclusion": "is zebra"},
    {"id": "R6", "conditions": ["is mammal", "long neck", "long legs"], "conclusion": "is giraffe"},
    {"id": "R7", "conditions": ["is mammal", "is carnivore", "black stripes"], "conclusion": "is tiger"}
]

print("Knowledge Base Loaded")
```

```python
# Initial Observed Facts

working_memory = {
    "has hair",
    "eats meat"
}

print("Initial Working Memory:")
for fact in working_memory:
    print("-", fact)
```

```python
# Forward Chaining Inference Engine

fired_rules = []
identified_animals = ["is lion", "is tiger", "is zebra", "is giraffe"]

while True:

    conflict_set = []

    # Match Phase
    for rule in rules:
        if all(condition in working_memory for condition in rule["conditions"]):
            if rule["conclusion"] not in working_memory:
                conflict_set.append(rule)

    # Stop if no rule can fire
    if len(conflict_set) == 0:
        break

    # Select first rule
    selected_rule = conflict_set[0]

    # Fire rule
    working_memory.add(selected_rule["conclusion"])
    fired_rules.append(selected_rule["id"])

    print(selected_rule["id"], "Fired →", selected_rule["conclusion"])

    # Stop when an animal is identified
    if selected_rule["conclusion"] in identified_animals:
        break
```

```python
# Display Final Result

print("\nFinal Working Memory:")
for fact in sorted(working_memory):
    print("-", fact)

print("\nFiring Sequence:")
print(" → ".join(fired_rules))

print("\nIdentified Animal:")

animal = None
for item in ["is lion", "is tiger", "is zebra", "is giraffe"]:
    if item in working_memory:
        animal = item
        break

if animal:
    print(animal.replace("is ", "").title())
elif "is carnivore" in working_memory:
    print("Carnivore (Most Specific Class)")
elif "is mammal" in working_memory:
    print("Mammal (Most Specific Class)")
else:
    print("No Identification Possible")
```

```python
# Exercise 1: Identify Tiger, Zebra and Giraffe

# Rule order updated so Tiger is checked before Lion
rules_ex1 = [
    {"id": "R1", "conditions": ["has hair"], "conclusion": "is mammal"},
    {"id": "R2", "conditions": ["gives milk"], "conclusion": "is mammal"},
    {"id": "R3", "conditions": ["eats meat"], "conclusion": "is carnivore"},
    {"id": "R7", "conditions": ["is mammal", "is carnivore", "black stripes"], "conclusion": "is tiger"},
    {"id": "R4", "conditions": ["is mammal", "is carnivore"], "conclusion": "is lion"},
    {"id": "R5", "conditions": ["is mammal", "black stripes"], "conclusion": "is zebra"},
    {"id": "R6", "conditions": ["is mammal", "long neck", "long legs"], "conclusion": "is giraffe"}
]

# Forward Chaining Function
def forward_chain(facts):
    wm = set(facts)
    fired = []
    animals = ["is tiger", "is lion", "is zebra", "is giraffe"]

    while True:
        conflict_set = []

        for rule in rules_ex1:
            if all(c in wm for c in rule["conditions"]) and rule["conclusion"] not in wm:
                conflict_set.append(rule)

        if not conflict_set:
            break

        rule = conflict_set[0]
        wm.add(rule["conclusion"])
        fired.append(rule["id"])

        if rule["conclusion"] in animals:
            break

    animal = "No Identification"
    for a in animals:
        if a in wm:
            animal = a.replace("is ", "").title()

    return fired, animal

# Test Cases
tests = {
    "Tiger": {"has hair", "eats meat", "black stripes"},
    "Zebra": {"has hair", "black stripes"},
    "Giraffe": {"has hair", "long neck", "long legs"}
}

for name, facts in tests.items():
    fired, animal = forward_chain(facts)
    print(f"\n{name}")
    print("Facts:", ", ".join(sorted(facts)))
    print("Firing Sequence:", " → ".join(fired))
    print("Identified Animal:", animal)
```

```python
# Exercise 3: Fire ALL applicable rules per cycle

working_memory = {
    "has hair",
    "eats meat",
    "black stripes"
}

fired_rules = []

while True:

    conflict_set = []

    # Find all applicable rules
    for rule in rules:
        if all(condition in working_memory for condition in rule["conditions"]):
            if rule["conclusion"] not in working_memory:
                conflict_set.append(rule)

    if len(conflict_set) == 0:
        break

    print("Conflict Set:", [rule["id"] for rule in conflict_set])

    # Fire ALL rules in this cycle
    for rule in conflict_set:
        working_memory.add(rule["conclusion"])
        fired_rules.append(rule["id"])
        print(rule["id"], "Fired →", rule["conclusion"])

print("\nFinal Working Memory:")
for fact in sorted(working_memory):
    print("-", fact)

print("\nFiring Sequence:")
print(" → ".join(fired_rules))
```
